```python
# import numpy
import numpy as np
```

```python
# creating arrays with zeros, ones and constants
zeros_array = np.zeros((3,3))
ones_array = np.ones((3,3))
constant_array = np.full((3,3),5)
zeros_array,ones_array,constant_array
```

```python
# converting list to array
arr = np.array([1,2,3])
arr
```

```python
# generating ranges of numbers
range_array = np.arange(10)
range_array
```

```python
# creating arrays with linear spacing
linspace_array = np.linspace(0,1,num=5)
linspace_array
```

```python
# mutlti-dimesional arrays
zeros_matrix = np.zeros((3,3))
```

```python
zeros_matrix
```

```python
# indexing and slicing
arr = np.array([[2,3,4],[4,5,6]])
# get the first row
arr[0]
# get first col
arr[:,1]
```

```python
# Generating random arrays
# to ensure reproducibility, we set random_seed
np.random.seed(2)
rand_array = np.random.rand(3,3)
rand_array
```

```python
# for random numbers from a normal distribution or integers within a range
normal_distribution = np.random.randn(5,2)
random_integers = np.random.randint(0,10,size=(5,2))
random_integers,normal_distribution
```

```python
# elemnt wise operations
zeros_array = np.zeros(4)
constant_array = np.full(4,4)
zeros_array + constant_array
```

```python
# comparsion operations
arr= np.array([1,2,3,4])
greater_than_2 = arr>2
greater_than_2
```

```python
# select elements based on conditons
selected_elements = arr[arr>1]
selected_elements
```

```python
min = arr.min()
max = arr.max()
sum = arr.sum()
mean = arr.mean()
standard_deviation = arr.std()
min,max,sum,mean,standard_deviation
```
